## Carregando o modelo

```python
import truststore
truststore.inject_into_ssl()   # para o download passar pelo certificado da empresa

import torch
from transformers import MarianMTModel, MarianTokenizer

NOME_MODELO = "Helsinki-NLP/opus-mt-en-ROMANCE"
tokenizer = MarianTokenizer.from_pretrained(NOME_MODELO)
modelo = MarianMTModel.from_pretrained(NOME_MODELO)
modelo.eval()

print("GPU disponível:", torch.cuda.is_available())
```

## Tradução em lotes

```python
def traduzir_lote(textos):
    entrada = [">>pt<< " + t for t in textos]
    batch = tokenizer(entrada, return_tensors="pt", padding=True,
                      truncation=True, max_length=512)
    with torch.no_grad():
        saida = modelo.generate(**batch, num_beams=2, max_new_tokens=512)
    return tokenizer.batch_decode(saida, skip_special_tokens=True)

# Teste com 3 textos
amostra = df_silver["feedbackDoCliente"].head(3).tolist()
for en, pt in zip(amostra, traduzir_lote(amostra)):
    print("EN:", en)
    print("PT:", pt)
    print()
```

### Traduzindo tudo em lotes

```python
import os
from tqdm import tqdm

COL_ORIGEM, COL_DESTINO = "feedbackDoCliente", "feedbackDoClientePT"
CHECKPOINT = "feedback_hf_checkpoint.csv"
TAM_LOTE = 16

if os.path.exists(CHECKPOINT):
    df_silver[COL_DESTINO] = pd.read_csv(CHECKPOINT, encoding="utf-8-sig")[COL_DESTINO]
elif COL_DESTINO not in df_silver.columns:
    df_silver[COL_DESTINO] = None

pendentes = df_silver.index[df_silver[COL_DESTINO].isna()].tolist()
print("Pendentes:", len(pendentes))

for k in tqdm(range(0, len(pendentes), TAM_LOTE), desc="Traduzindo"):
    idx = pendentes[k:k + TAM_LOTE]
    textos = df_silver.loc[idx, COL_ORIGEM].tolist()
    df_silver.loc[idx, COL_DESTINO] = traduzir_lote(textos)

    if (k // TAM_LOTE) % 10 == 0:   # checkpoint a cada 10 lotes
        df_silver[[COL_DESTINO]].to_csv(CHECKPOINT, index=False, encoding="utf-8-sig")

df_silver[[COL_DESTINO]].to_csv(CHECKPOINT, index=False, encoding="utf-8-sig")
```

# Análise de sentimentos utilizando Leia

```python
from leia import SentimentIntensityAnalyzer
analyzer = SentimentIntensityAnalyzer()

def analisar_leia(texto):
    if pd.isna(texto) or str(texto).strip() == "":
        return pd.Series([0.0, "Neutro"])

    scores = analyzer.polarity_scores(str(texto))
    compound = scores["compound"]

    if compound >= 0.05:
        sentimento = "Positivo"
    elif compound <= -0.05:
        sentimento = "Negativo"
    else:
        sentimento = "Neutro"

    return pd.Series([compound, sentimento])

#aplica na coluna traduzida em português
df_gold[["score_sentimento", "sentimento"]] = df_gold["feedbackDoCliente"].apply(analisar_leia)

df_gold.to_csv("df_gold.csv", index=False, sep=",", decimal=".", encoding="utf-8-sig")
```

Com base nesse perfil, o próximo passo será analisar a taxa de cancelamento (*churn*) entre os clientes que apresentam essas características, a fim de compreender o impacto dessa combinação de fatores na evasão de clientes.

```python

# Filtrar clientes com as características desejadas
filtro = df[
    (df['dependentes'] == False) &
    (df['conjuge'] == False) &
    (df['contrato'] == 'Month-to-month') &
    (df['serviçoDeInternet'] == 'Fiber optic')
]

# Calcular quantidade de clientes com esse perfil
total_clientes = len(filtro)

# Calcular quantidade de cancelamentos desse perfil
cancelamentos = filtro['cancelamento'].sum()

# Calcular taxa de churn do perfil
taxa_churn = (cancelamentos / total_clientes) * 100 if total_clientes > 0 else 0.0

# Calcular total de cancelamentos da base inteira
total_cancelamentos = df['cancelamento'].sum()

# Calcular representatividade nos cancelamentos totais
representatividade_cancelamentos = (
    (cancelamentos / total_cancelamentos) * 100
    if total_cancelamentos > 0 else 0.0
)

# Exibir resultados
print(f'Total de clientes com este perfil: {total_clientes}')
print(f'Cancelamentos deste perfil: {cancelamentos}')
print(f'Taxa de churn do perfil: {taxa_churn:.2f}%')
print(f'Representatividade nos cancelamentos totais: {representatividade_cancelamentos:.2f}%')
```

## Resumo da Análise de Sentimentos

* **Objetivo:** Classificar automaticamente a polaridade dos feedbacks dos clientes (`feedbackDoCliente`) para medir o nível de satisfação geral e identificar riscos de cancelamento.
* **Variáveis Geradas (`df_gold`):**
  * `score_sentimento`: Pontuação contínua de polaridade entre **$-1$** (muito negativo) e **$+1$** (muito positivo).
  * `sentimento`: Classificação categórica em **Positivo** ($\ge 0.05$), **Neutro** (entre $-0.05$ e $+0.05$) e **Negativo** ($\le -0.05$).
* **Principais Cruzamentos Analisados:**
  * **Sentimento $\times$ Cancelamento (`cancelamento`):** Evidencia tanto o cancelamento por insatisfação quanto o *"churn silencioso"* (clientes com feedbacks positivos
  * **Sentimento $\times$ Contrato (`contrato`):** Compara o nível de satisfação entre planos mensais e contratos de longo prazo.
  * **Sentimento $\times$ Mensalidade (`mensalidade`):** Quantifica a receita mensal perdida e o valor financeiro dos clientes ativos que estão insatisfeitos.
