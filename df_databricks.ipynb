## Função para tradução de colunas do dataframe

```python
def traduzir(texto, tentativas=3):
    if not isinstance(texto, str) or not texto.strip():
        return texto
    
    for i in range(tentativas):
        try:
            # Cria o tradutor dentro da função
            return GoogleTranslator(source="en", target="pt").translate(texto)
        except Exception:
            time.sleep(1.5 * (i + 1))   # espera um pouco mais a cada falha

    return None 
```

from deep_translator import GoogleTranslator
from concurrent.futures import ThreadPoolExecutor
from tqdm import tqdm

# 1. Prepara os textos válidos da coluna 'feedbackDoCliente'
coluna_alvo = 'feedbackDoCliente'
coluna_nova = 'feedbackDoCliente_pt'

mascara_validos = df[coluna_alvo].notna() & (df[coluna_alvo].astype(str).str.strip() != '')
textos = df.loc[mascara_validos, coluna_alvo].astype(str).tolist()

# 2. Função que traduz um texto por vez com tratamento de erro
def traduzir_texto(texto):
    try:
        # Cria uma instância do tradutor por thread para evitar conflitos
        return GoogleTranslator(source='en', target='pt').translate(texto)
    except Exception:
        # Se falhar por oscilação de rede, retorna o texto original
        return texto

# 3. Executa em paralelo (10 threads simultâneas aceleram muito o processo)
with ThreadPoolExecutor(max_workers=10) as executor:
    traducoes = list(tqdm(
        executor.map(traduzir_texto, textos),
        total=len(textos),
        desc="Traduzindo feedbacks"
    ))

# 4. Salva o resultado na camada Gold do DataFrame
df.loc[mascara_validos, coluna_nova] = traducoes
